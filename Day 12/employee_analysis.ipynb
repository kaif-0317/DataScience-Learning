```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the cleaned employee dataset
df = pd.read_csv("../Day 10/cleaned_employee_data.csv")

# Display the first few rows
df.head()
```

## 7. Business Insights

The analysis provides the following key business insights:

- Total salary paid to employees
- Average employee salary
- Highest employee salary
- Lowest employee salary
- Total number of employees
- Employee distribution across departments

```python
# Count employees in each department
department_counts = df["Department"].value_counts()

print(department_counts)

# Create a bar chart for department distribution
plt.figure(figsize=(8, 5))

department_counts.plot(kind="bar")

plt.title("Employee Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
plt.show()
```

## 6. Employees by Department

This analysis shows the number of employees in each department.

```python
# Create a bar chart to compare employee salaries
plt.figure(figsize=(8, 5))

plt.bar(df["Name"], df["Salary"])

plt.title("Employee Salary Comparison")
plt.xlabel("Employee Name")
plt.ylabel("Salary (₹)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Salary by Employee

The following bar chart compares the salaries of individual employees.

```python
# Calculate key business metrics
total_salary = df["Salary"].sum()
average_salary = df["Salary"].mean()
highest_salary = df["Salary"].max()
lowest_salary = df["Salary"].min()
total_employees = len(df)

print(f"Total Employees   : {total_employees}")
print(f"Total Salary      : ₹{total_salary:,}")
print(f"Average Salary    : ₹{average_salary:,.2f}")
print(f"Highest Salary    : ₹{highest_salary:,}")
print(f"Lowest Salary     : ₹{lowest_salary:,}")
```
